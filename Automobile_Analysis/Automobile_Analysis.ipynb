```python
import pandas as pd
import numpy as np
df=pd.read_csv("automobile_dataset.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
(df.isnull().sum()/len(df))*100
```

```python
df["price"]=df["price"].fillna(df["price"]).median()
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
df["peak-rpm"] = df["peak-rpm"].fillna(df["peak-rpm"].mean())
df["stroke"] = df["stroke"].fillna(df["stroke"].mean())
df["bore"] = df["bore"].fillna(df["bore"].mean())
df["num-of-doors"] = df["num-of-doors"].fillna(df["num-of-doors"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df = df.drop(columns=["normalized-losses"])
```

```python
df["mileage_category"] = np.where(df["city-mpg"] > 20, "High", "Low")
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.histplot(df["price"], kde=True)
plt.title("Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["horsepower"])
plt.title("Horsepower Outliers")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x=df["body-style"])
plt.title("Body Style Count")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x=df["horsepower"], y=df["price"])
plt.title("Horsepower vs Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.regplot(x="engine-size", y="price", data=df)
plt.title("Engine Size vs Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="fuel-type", y="price", data=df)
plt.title("Fuel Type vs Price")
plt.show()
```
